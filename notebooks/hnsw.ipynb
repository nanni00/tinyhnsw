# Create and Visualize the HNSW

Questions: 
- Node X has some connections in layer L: are these connections preserved also in layers L-1, L-2, ...? Why?
- Do the parameters M, M_max, etc, have any influence on that?

```python
import math
from tinyhnsw.visualization import visualize_hnsw_index, generate_tiny_world

config = DEFAULT_CONFIG
config.M_max = 4
config.M_max0 = 6
config.M = 4
config.m_L = (1. / math.log(4))

total = 20
layout = generate_tiny_world(total, 50, 50)
vectors = np.array([layout[ix] for ix in range(total)])
index = HNSWIndex(d=2, config=config)
index.add(vectors)

visualize_hnsw_index(index, layout)
```

# Benchmark HNSW with different configurations

### Load the SIFT10K/SIFT1M Dataset

```python
from tinyhnsw.utils import load_sift, evaluate, load_sift_1M

# data, queries, labels = load_sift()
data, queries, labels = load_sift_1M()
```

```python
import faiss
```

```python
N = 200_000
limit_data = data[:N]
limit_queries = queries[labels < N]
limit_labels = labels[labels < N]

faiss_index = faiss.IndexHNSWFlat(128, 8)
faiss_index.train(limit_data)
faiss_index.is_trained
faiss_index.add(limit_data)
```

```python
params = faiss.SearchParametersHNSW(efSearch=200)
D, I = faiss_index.search(limit_queries, k=1, params=params)     # actual search
```

```python
import numpy as np
sum(limit_labels == I.reshape(1, -1)[0]) * 100 / limit_labels.shape[0]
```
